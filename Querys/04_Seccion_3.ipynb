Las reservas sin hijos tienen una tasa de cancelación de 37.22%, mientras que las reservas con hijos registran un 34.92%. Se observa que los clientes que viajan con hijos cancelan un poco menos sus reservas.

```python
%sql

SELECT
    CASE
        WHEN COALESCE(children, 0) > 0
          OR COALESCE(babies, 0) > 0
        THEN 'Con Hijos'
        ELSE 'Sin Hijos'
    END AS tipo_reserva,

    COUNT(*) AS total_reservas,

    AVG(is_canceled) * 100 AS tasa_cancelacion

FROM cursodataacademy.hoteles.hotel_bookings

GROUP BY
    CASE
        WHEN COALESCE(children, 0) > 0
          OR COALESCE(babies, 0) > 0
        THEN 'Con Hijos'
        ELSE 'Sin Hijos'
    END

ORDER BY tasa_cancelacion DESC;
```

Las reservas que no tienen solicitudes especiales presentan una tasa de cancelación de 47.72%, siendo la más alta. En cambio, cuando los clientes realizan más solicitudes, el porcentaje generalmente disminuye, por lo que parece existir una relación entre las solicitudes y una mayor intención de mantener la reserva.

```python
%sql

SELECT
    total_of_special_requests,

    COUNT(*) AS total_reservas,

    AVG(is_canceled) * 100 AS tasa_cancelacion

FROM cursodataacademy.hoteles.hotel_bookings

GROUP BY total_of_special_requests

ORDER BY total_of_special_requests ASC;
```

Los clientes que realizaron entre 1 y 3 cambios presentan la menor tasa de cancelación, con un 15.59%. Por otro lado, las reservas sin modificaciones alcanzan el 40.85%, siendo el grupo con mayor porcentaje de cancelaciones.

```python
%sql

SELECT
    CASE
        WHEN booking_changes = 0 THEN 'Sin Cambios'
        WHEN booking_changes BETWEEN 1 AND 3 THEN '1 a 3 cambios'
        WHEN booking_changes > 3 THEN 'Más de 3 cambios'
        ELSE 'Sin información'
    END AS categoria_cambios,

    COUNT(*) AS total_reservas,

    AVG(is_canceled) * 100 AS tasa_cancelacion

FROM cursodataacademy.hoteles.hotel_bookings

GROUP BY
    CASE
        WHEN booking_changes = 0 THEN 'Sin Cambios'
        WHEN booking_changes BETWEEN 1 AND 3 THEN '1 a 3 cambios'
        WHEN booking_changes > 3 THEN 'Más de 3 cambios'
        ELSE 'Sin información'
    END

ORDER BY tasa_cancelacion ASC;
```

Resort Hotel tiene un precio promedio de USD 105.79 en sus reservas canceladas, ligeramente superior a City Hotel, que registra USD 104.69. Sin embargo, City Hotel presenta una cantidad mucho mayor de cancelaciones, con 33,102 frente a las 11,122 de Resort Hotel.
